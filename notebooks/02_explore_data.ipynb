```python
import json

with open('../data/raw/cuad-main/data/CUADv1.json', 'r') as f:
    data = json.load(f)
```

```python
clean_data=[]

for contract in data['data']:
    contract_name=contract['title']
    for paragraph in contract['paragraphs']:
        for qa in paragraph['qas']:
            clause_type=qa['question']
            if qa['answers']:
                clause_text=qa['answers'][0]['text']
                clean_data.append({ "contract":contract_name, "clause_type":clause_type, "clause_text":clause_text })
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(clean_data)
```
