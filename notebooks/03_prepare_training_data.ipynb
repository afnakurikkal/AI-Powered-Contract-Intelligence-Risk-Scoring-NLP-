```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../data/processed/cuad_clean.csv")
df["label"] = df["clause_type"].str.extract(r'related to "([^"]+)"')
df = df[["contract", "clause_text", "label"]]
```

```python
contracts = df["contract"].unique()
train_c, temp_c = train_test_split(contracts, test_size=0.2, random_state=42)
val_c, test_c = train_test_split(temp_c, test_size=0.5, random_state=42)
```

```python
df[df.contract.isin(train_c)].to_csv("../data/processed/train.csv", index=False)
df[df.contract.isin(val_c)].to_csv("../data/processed/val.csv", index=False)
df[df.contract.isin(test_c)].to_csv("../data/processed/test.csv", index=False)
print(df.label.nunique(), "labels")
```
